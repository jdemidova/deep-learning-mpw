# Abstract
## Mandatory tasks
> a) Start from **simple CNN** architectures and **progressively increase their complexity** to show
the benefit of depth.

This has been done by creating **N** different CNN models with increasing depth and complexity. The performance of each model is evaluated and compared to demonstrate the benefits of deeper architectures.
> b) Show the **importance of hyperparameter tuning**.

Hyperparameter tuning is performed for each model to optimize their performance. This includes adjusting parameters such as learning rate, batch size, and number of epochs. The impact of hyperparameter tuning on the model's performance is analyzed and described.
> c) Show models that **overfit and underfit** and explain the reasons for that.

Models that overfit and underfit are created by adjusting the complexity of the models and INPUT TEXT. The reasons for overfitting and underfitting are explained in terms of INPUT TEXT.
> d) Experiment with **regularization techniques**.

Regularization techniques such as dropout, L1/L2 regularization, and data augmentation are applied to the models to prevent overfitting. The effectiveness of these techniques is evaluated and compared.

> e) Experiment with **different optimization algorithms** (e.g., Adam, RMSprop) and **compare** their performance.

Different optimization algorithms are tested on the models. The performance of each optimizer is compared in terms of convergence speed and final accuracy.
